```python
from pathlib import Path
import warnings
warnings.filterwarnings("error", category=FutureWarning)
warnings.filterwarnings("error", category=DeprecationWarning)
ROOT = Path.cwd()
DATA = ROOT / "data"
print("Working root:", ROOT)
```

```python
import pandas as pd
import numpy as np
pd.set_option("display.max_columns", 20)
```

```python
import numpy as np
```

## 1. Ma trận số ThermoFlow

```python
df = pd.read_csv(DATA / "thermoflow" / "thermoflow_matrix.csv")
X = df.to_numpy(dtype=float)
print("shape =", X.shape, "; ndim =", X.ndim)
print("columns =", list(df.columns))
```

Với quy ước hàng là quan sát và cột là biến số, `X.shape == (300, 6)` có nghĩa 300 quan sát và 6 biến.

## 2. Lựa chọn và điều kiện trên cả mảng

```python
temp_out = X[:, 1]
mask = temp_out > np.quantile(temp_out, 0.95)
print("Số quan sát trên phân vị 95%:", mask.sum())
print(temp_out[mask][:5])
```

## 3. Axis là câu hỏi về chiều tổng hợp

```python
col_means = X.mean(axis=0)
row_means = X.mean(axis=1)
print("col_means shape:", col_means.shape)
print("row_means shape:", row_means.shape)
print("col means:", np.round(col_means, 3))
```

Không diễn giải `row_means` như một đại lượng vật lý: nó đang trung bình nhiệt độ, lưu lượng, áp suất, nồng độ và năng lượng - những đại lượng khác đơn vị.

## 4. Broadcasting để đổi đơn vị

```python
temperature_in_k = df["temperature_in_c"].to_numpy() + 273.15
print(temperature_in_k[:5])
```

### Kiểm tra bằng chứng

Một output có shape đúng vẫn có thể vô nghĩa nếu ta tổng hợp sai chiều hoặc trộn các đại lượng không cùng ý nghĩa.
