# Chương 15. Từ kết quả phân tích đến bằng chứng

**Câu hỏi mở đầu:** Khi nào một kết quả phân tích đủ mạnh để trở thành một nhận định?

Notebook này là tài nguyên đồng hành của chương. Mỗi phần đều đi theo nhịp **câu hỏi → dữ liệu → mã → kết quả → diễn giải → kiểm tra bằng chứng**.

## Mục tiêu

- Tái hiện các ví dụ cốt lõi của chương bằng mã có thể chạy lại.
- Kiểm tra giả định trước khi diễn giải output.
- Kết thúc bằng ít nhất một câu hỏi về điều mà kết quả **chưa** cho biết.

```python
from pathlib import Path
import warnings
warnings.filterwarnings("error", category=FutureWarning)
warnings.filterwarnings("error", category=DeprecationWarning)
ROOT = Path.cwd()
DATA = ROOT / "data"
print("Working root:", ROOT)
```

```python
import pandas as pd
import numpy as np
pd.set_option("display.max_columns", 20)
```

## 1. Output → phát hiện → nhận định

```python
tx=pd.read_csv(DATA / "metromart" / "metromart_transactions.csv")
stores=pd.read_csv(DATA / "metromart" / "metromart_stores.csv")
valid=tx["unit_price"].notna() & tx["discount_pct"].between(0,1)
work=tx.loc[valid].copy()
work["net_sales"]=work["unit_price"]*work["quantity"]*(1-work["discount_pct"])
work=work.merge(stores[["store_id","store_type","region","floor_area_m2"]],on="store_id",validate="many_to_one")
summary=work.groupby("store_type")["net_sales"].agg(["count","mean","median"])
display(summary)
```

**Output** là bảng trên. **Phát hiện** có thể là “trung bình giữa các loại cửa hàng khác nhau”. **Nhận định** phải thêm độ lớn, phạm vi và giới hạn.

## 2. Audit một nhận định quá mạnh

```python
claim="Cửa hàng loại A hiệu quả hơn và loại hình cửa hàng gây ra doanh thu cao hơn."
checks={
    "Có metric cụ thể?": False,
    "Có phạm vi dữ liệu?": False,
    "Có kiểm tra khác biệt về khu vực/quy mô?": False,
    "Có thiết kế nhân quả?": False,
}
print(claim)
for k,v in checks.items(): print(k, "->", v)
```

Một phiên bản phù hợp hơn: “Trong dữ liệu giao dịch quan sát được, một số loại cửa hàng có doanh thu giao dịch trung bình cao hơn. Tuy nhiên các nhóm cũng khác về khu vực và quy mô; kết quả mô tả mối liên hệ chứ chưa chứng minh nguyên nhân.”

## 3. Dấu vết phân tích

```python
trace=["metromart_transactions.csv","lọc discount hợp lệ","tạo net_sales","nối store metadata","group summary","nhận định có giới hạn"]
for i,step in enumerate(trace,1): print(i,step)
```

### Rà soát bằng chứng

Trước khi xuất bản một nhận định, kiểm tra: truy ngược được, liên quan câu hỏi, xử lý dữ liệu toàn vẹn, có độ lớn, có uncertainty khi cần, có giới hạn, ngôn ngữ tương xứng.
