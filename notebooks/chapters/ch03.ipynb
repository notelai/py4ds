## 2. Điều kiện như một quy tắc

```python
def eligible(record):
    return record["total_spend"] >= 1000 and record["visits"] >= 5
print(eligible(customer))
```

## 3. Vòng lặp và reason code

```python
customers = [
    {"customer_id":"C001","total_spend":1250,"visits":8},
    {"customer_id":"C002","total_spend":700,"visits":9},
    {"customer_id":"C003","total_spend":1300,"visits":3},
]
results=[]
for r in customers:
    if r["total_spend"] < 1000:
        reason="chi tiêu dưới ngưỡng"
    elif r["visits"] < 5:
        reason="số lần ghé dưới ngưỡng"
    else:
        reason="đủ điều kiện"
    results.append((r["customer_id"], reason))
print(results)
```

### Kiểm tra bằng chứng

Output “không đủ điều kiện” phản ánh quy tắc ta đặt ra, không phải một đánh giá khách quan về giá trị của khách hàng.
