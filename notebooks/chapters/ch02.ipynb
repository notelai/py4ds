## 1. Đại lượng, kiểu dữ liệu và đơn vị

```python
unit_price = 25.0
quantity = 8
discount_pct = 0.10
gross_sales = unit_price * quantity
net_sales = gross_sales * (1 - discount_pct)
print(type(unit_price), type(quantity), type(discount_pct))
print(f"Gross sales = {gross_sales:.2f}")
print(f"Net sales   = {net_sales:.2f}")
```

Một con số chỉ có ý nghĩa khi biết đại lượng và đơn vị. `25.0` có thể là nghìn đồng, USD hoặc một đại lượng khác.

## 2. Khi kiểu dữ liệu không phù hợp

```python
discount_text = "15%"
try:
    unit_price * quantity * (1 - discount_text)
except TypeError as exc:
    print(type(exc).__name__, "->", exc)
```

```python
discount_pct = float(discount_text.strip("%")) / 100
print("discount_pct =", discount_pct)
print("net_sales =", unit_price * quantity * (1 - discount_pct))
```

## 3. Phần trăm thay đổi và trường hợp biên

```python
def percent_change(old, new):
    if old == 0:
        raise ValueError("old phải khác 0 để phần trăm thay đổi được xác định")
    return (new - old) / old * 100
print(percent_change(100, 125))
try:
    print(percent_change(0, 10))
except ValueError as exc:
    print("Kiểm tra đầu vào:", exc)
```

### Kiểm tra bằng chứng

Một biểu thức chạy được chưa đủ. Hãy kiểm tra đơn vị, ý nghĩa của biến và miền giá trị hợp lệ trước khi dùng kết quả trong nhận định.
