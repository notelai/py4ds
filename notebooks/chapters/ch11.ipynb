```python
from pathlib import Path
import warnings
warnings.filterwarnings("error", category=FutureWarning)
warnings.filterwarnings("error", category=DeprecationWarning)
ROOT = Path.cwd()
DATA = ROOT / "data"
print("Working root:", ROOT)
```

```python
import pandas as pd
import numpy as np
pd.set_option("display.max_columns", 20)
```

```python
import numpy as np
```

## 1. Trung tâm và độ phân tán

```python
tx=pd.read_csv(DATA / "metromart" / "metromart_transactions.csv")
valid=tx["discount_pct"].between(0,1) & tx["unit_price"].notna()
x=(tx.loc[valid,"unit_price"]*tx.loc[valid,"quantity"]*(1-tx.loc[valid,"discount_pct"])).to_numpy()
print("mean:", x.mean())
print("median:", np.median(x))
print("Q1,Q3:", np.quantile(x,[.25,.75]))
print("std sample:", np.std(x,ddof=1))
```

## 2. Cùng trung bình, khác cấu trúc

```python
a=np.array([8,9,10,11,12],dtype=float)
b=np.array([0,10,10,10,20],dtype=float)
print("means:",a.mean(),b.mean())
print("stds:",np.std(a,ddof=1),np.std(b,ddof=1))
```

## 3. NumPy và pandas: làm rõ ddof

```python
s=pd.Series(a)
print(np.std(a,ddof=1), s.std(ddof=1))
```
