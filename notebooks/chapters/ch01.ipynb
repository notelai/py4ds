**Câu hỏi mở đầu:** Một bảng dữ liệu bán hàng thực sự có thể giúp ta trả lời những câu hỏi nào?

Notebook này là tài nguyên đồng hành của chương. Mỗi phần đều đi theo nhịp **câu hỏi → dữ liệu → mã → kết quả → diễn giải → kiểm tra bằng chứng**.

## Mục tiêu

- Tái hiện các ví dụ cốt lõi của chương bằng mã có thể chạy lại.
- Kiểm tra giả định trước khi diễn giải output.
- Kết thúc bằng ít nhất một câu hỏi về điều mà kết quả **chưa** cho biết.

```python
from pathlib import Path
import warnings
warnings.filterwarnings("error", category=FutureWarning)
warnings.filterwarnings("error", category=DeprecationWarning)
ROOT = Path.cwd()
DATA = ROOT / "data"
print("Working root:", ROOT)
```

```python
import pandas as pd
import numpy as np
pd.set_option("display.max_columns", 20)
```

## 1. Một bảng nhỏ có thể trả lời câu hỏi nào?

```python
df = pd.read_csv(DATA / "metromart" / "metromart_small.csv")
print(df.shape)
display(df.head(8))
```

Một hàng trong bảng này là **một giao dịch nhỏ**. Vì vậy câu hỏi ở cấp khách hàng hoặc nguyên nhân mua hàng vượt quá đơn vị phân tích hiện có.

```python
product_frequency = df["product"].value_counts()
store_quantity = df.groupby("store")["quantity"].sum().sort_values(ascending=False)
print("Sản phẩm xuất hiện nhiều nhất:")
display(product_frequency.head())
print("Tổng số lượng theo cửa hàng:")
display(store_quantity)
```

### Kiểm tra bằng chứng

“Sản phẩm xuất hiện nhiều nhất” không đồng nghĩa “sản phẩm được yêu thích nhất”. Dữ liệu chỉ ghi giao dịch, không ghi sở thích hoặc động cơ.

## 2. Từ câu hỏi đến biến

```python
question_map = {
    "Sản phẩm nào xuất hiện nhiều nhất?": ["product"],
    "Cửa hàng nào có tổng số lượng cao nhất?": ["store", "quantity"],
    "Vì sao khách hàng mua nhiều ở cửa hàng đó?": None,
}
for q, cols in question_map.items():
    print(q, "->", cols if cols else "Chưa đủ biến để trả lời")
```

1. Viết hai câu hỏi mô tả và một câu hỏi so sánh từ bảng.

3. Viết một câu hỏi mà dữ liệu hiện tại **không thể** trả lời và nêu biến cần bổ sung.
