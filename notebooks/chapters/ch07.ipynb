## Mục tiêu

- Tái hiện các ví dụ cốt lõi của chương bằng mã có thể chạy lại.
- Kiểm tra giả định trước khi diễn giải output.
- Kết thúc bằng ít nhất một câu hỏi về điều mà kết quả **chưa** cho biết.

```python
from pathlib import Path
import warnings
warnings.filterwarnings("error", category=FutureWarning)
warnings.filterwarnings("error", category=DeprecationWarning)
ROOT = Path.cwd()
DATA = ROOT / "data"
print("Working root:", ROOT)
```

```python
import pandas as pd
import numpy as np
pd.set_option("display.max_columns", 20)
```

```python
df = pd.read_csv(DATA / "metromart" / "metromart_sales_raw.csv")
```

## 1. Câu hỏi → biến → quan sát

```python
subset = df.loc[:, ["transaction_id","category","unit_price","quantity","discount_pct","transaction_type"]]
display(subset.head())
```

## 2. Điều kiện lọc

```python
candidate = df.loc[(df["unit_price"] > 100) & (df["quantity"] < 5),
                   ["transaction_id","category","unit_price","quantity","transaction_type"]]
print("candidate rows:", len(candidate))
display(candidate.head())
```

Ngưỡng 100 và 5 chỉ là một định nghĩa làm việc. Nếu đổi ngưỡng, danh sách có thể thay đổi mạnh.

## 3. Biến dẫn xuất

```python
valid_discount = df["discount_pct"].between(0,1)
work = df.loc[valid_discount & df["unit_price"].notna()].copy()
work["net_sales"] = work["unit_price"] * work["quantity"] * (1 - work["discount_pct"])
display(work[["unit_price","quantity","discount_pct","net_sales"]].head())
```

### Kiểm tra bằng chứng

Danh sách được lọc chỉ đúng theo định nghĩa đã chọn. Hãy kiểm tra độ nhạy của kết luận khi ngưỡng thay đổi.
