Notebook này là tài nguyên đồng hành của chương. Mỗi phần đều đi theo nhịp **câu hỏi → dữ liệu → mã → kết quả → diễn giải → kiểm tra bằng chứng**.

## Mục tiêu

- Kết thúc bằng ít nhất một câu hỏi về điều mà kết quả **chưa** cho biết.

```python
from pathlib import Path
import warnings
warnings.filterwarnings("error", category=FutureWarning)
warnings.filterwarnings("error", category=DeprecationWarning)
ROOT = Path.cwd()
DATA = ROOT / "data"
print("Working root:", ROOT)
```

## 1. Từ mã lặp lại đến hàm

```python
def gross_sales(unit_price, quantity):
    """Tính doanh thu trước giảm giá."""
    return unit_price * quantity

def net_sales(unit_price, quantity, discount_pct=0.0):
    if not 0 <= discount_pct <= 1:
        raise ValueError("discount_pct phải nằm trong [0, 1]")
    return gross_sales(unit_price, quantity) * (1 - discount_pct)

print(net_sales(25, 8, 0.10))
```

## 2. `print` khác `return`

```python
def show_value(x):
    print(x)

def return_value(x):
    return x

a = show_value(10)
b = return_value(10)
print("a =", a, "; b =", b)
```

## 3. Hàm và giả định

```python
for d in [0.15, 1.5]:
    try:
        print("discount", d, "->", net_sales(100, 2, d))
    except ValueError as exc:
        print("discount", d, "->", exc)
```

### Kiểm tra bằng chứng

Hàm chạy không báo lỗi chỉ cho biết các kiểm tra đã viết được thỏa. Nó không chứng minh dữ liệu đầu vào đúng về nghiệp vụ nếu ta chưa mã hóa các điều kiện đó.
