```python
def percent_change(old,new):
    if old==0:
        raise ValueError("Giá trị ban đầu phải khác 0")
    return (new-old)/old*100
assert percent_change(100,125)==25
print("Mức thay đổi:", percent_change(100,125), "%")
try:
    percent_change(0,10)
except ValueError as e:
    print("Kiểm tra giả định:", e)
```
