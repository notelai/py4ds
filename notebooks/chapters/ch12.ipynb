```python
from pathlib import Path
import warnings
warnings.filterwarnings("error", category=FutureWarning)
warnings.filterwarnings("error", category=DeprecationWarning)
ROOT = Path.cwd()
DATA = ROOT / "data"
print("Working root:", ROOT)
```

```python
import pandas as pd
import numpy as np
pd.set_option("display.max_columns", 20)
```

```python
import matplotlib.pyplot as plt
plt.rcParams["figure.dpi"] = 100
```

```python
tx=pd.read_csv(DATA / "metromart" / "metromart_transactions.csv")
stores=pd.read_csv(DATA / "metromart" / "metromart_stores.csv")
valid=tx["discount_pct"].between(0,1) & tx["unit_price"].notna()
work=tx.loc[valid].copy()
work["net_sales"]=work["unit_price"]*work["quantity"]*(1-work["discount_pct"])
work=work.merge(stores[["store_id","store_type"]],on="store_id",validate="many_to_one")
```

## 1. Phân phối

```python
fig,ax=plt.subplots(figsize=(6,3.5))
ax.hist(work["net_sales"],bins=30)
ax.set_xlabel("Doanh thu sau giảm giá")
ax.set_ylabel("Số giao dịch")
ax.set_title("Phân phối doanh thu giao dịch")
plt.show()
```

## 2. So sánh nhóm bằng biểu đồ hộp

```python
groups=[g["net_sales"].to_numpy() for _,g in work.groupby("store_type")]
labels=[name for name,_ in work.groupby("store_type")]
fig,ax=plt.subplots(figsize=(6,3.5))
ax.boxplot(groups,tick_labels=labels,showfliers=False)
ax.set_ylabel("Doanh thu sau giảm giá")
ax.set_title("Phân phối theo loại cửa hàng")
plt.show()
```

## 3. Quan hệ hai biến

```python
sample=work.sample(700,random_state=7)
fig,ax=plt.subplots(figsize=(5.5,3.5))
ax.scatter(sample["unit_price"],sample["quantity"],s=12,alpha=.45)
ax.set_xlabel("Đơn giá")
ax.set_ylabel("Số lượng")
ax.set_title("Đơn giá và số lượng")
plt.show()
```
