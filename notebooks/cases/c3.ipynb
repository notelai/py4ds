## Câu hỏi

Mức hài lòng khác nhau giữa các khu vực đến đâu, và mẫu khảo sát có đại diện đủ để suy rộng không?

```python
pop=pd.read_csv(DATA/"citypulse"/"citypulse_population.csv")
rand=pd.read_csv(DATA/"citypulse"/"citypulse_random_sample.csv")
web=pd.read_csv(DATA/"citypulse"/"citypulse_web_sample.csv")
districts=pd.read_csv(DATA/"citypulse"/"citypulse_districts.csv")
for name,df in [("population",pop),("random",rand),("web",web)]:
    print(name,df["satisfied"].eq("yes").mean())
```

```python
by_district=(rand.assign(satisfied_num=rand["satisfied"].eq("yes"))
             .groupby("district_id")["satisfied_num"].agg(["count","mean"]).reset_index()
             .merge(districts,on="district_id",validate="one_to_one"))
display(by_district[["district_name","count","mean","population_size"]])
```

### Kiểm tra bằng chứng

Số người không hài lòng phải được đọc cùng mẫu số. Khác biệt nhóm trong một khảo sát quan sát cũng chưa cho phép kết luận nguyên nhân quản lý.
