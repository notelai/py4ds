## Câu hỏi

Chế độ vận hành nào cho mức tăng nhiệt độ ổn định hơn, sau khi loại các bản ghi được xác nhận không hợp lệ?

```python
df=pd.read_csv(DATA/"thermoflow"/"thermoflow_runs_raw.csv")
print(df.shape)
print("negative pressure",(df["pressure_kpa"]<0).sum())
print("calibration",df["sensor_status"].eq("CALIBRATION").sum())
```

```python
valid=(df["pressure_kpa"]>0) & ~df["sensor_status"].eq("CALIBRATION") & df["temperature_in_c"].notna() & df["temperature_out_c"].notna()
work=df.loc[valid].copy()
work["delta_t"]=work["temperature_out_c"]-work["temperature_in_c"]
summary=work.groupby("operating_mode")["delta_t"].agg(["count","mean","median","std"])
display(summary)
```

Một nhiệt độ cao có thể hiếm nhưng hợp lý trong Mode C. Quy tắc vật lý và thống kê phải được dùng cùng nhau, không thay thế nhau.
