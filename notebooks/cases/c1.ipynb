## Câu hỏi

Những loại cửa hàng nào đóng góp doanh thu lớn, và kết quả đó có ổn định giữa các khu vực không?

```python
tx=pd.read_csv(DATA/"metromart"/"metromart_transactions.csv")
stores=pd.read_csv(DATA/"metromart"/"metromart_stores.csv")
valid=tx["unit_price"].notna() & tx["discount_pct"].between(0,1)
work=tx.loc[valid].copy()
work["net_sales"]=work["unit_price"]*work["quantity"]*(1-work["discount_pct"])
work=work.merge(stores,on="store_id",validate="many_to_one")
print(work.shape)
print(work.isna().mean().sort_values(ascending=False).head())
```

```python
summary=work.groupby(["region","store_type"])["net_sales"].agg(["count","median","mean","sum"]).reset_index()
display(summary.head(12))
```

```python
pivot=summary.pivot(index="region",columns="store_type",values="sum")
display(pivot)
```
